```python
import yfinance as yf
import pandas as pd
```

```python
stocks = ['INE002A01018', 'INE00WC01027', 'INE018E01016', 'INE01B501018', 
          'INE020S01012', 'INE02ID01020', 'INE040A01034', 'INE075A01022', 
          'INE086A01029', 'INE08U801020', 'INE094S01041', 'INE09VQ01012', 
          'INE0BS701011', 'INE0BY001018', 'INE0GSL01016', 'INE0GSL01024', 
          'INE150G01020', 'INE237A01028', 'INE301A01014', 'INE377Y01014', 
          'INE483C01032', 'INE559R01029', 'INE575P01011', 'INE591D01014', 
          'INE633B01018', 'INE680Z01018', 'INE717W01049', 'INE732I01013', 
          'INE758E01017', 'INE835B01035', 'INE974Z01015', 
          'INF204KB15V2',
          'INE943C01027', 'INF109KC19V3',
          ]
```

```python
# get latest stock price for 1st stock as a sample
ticker = yf.Ticker(stocks[0])
latest_price = ticker.history(period="1d")['Close'].iloc[-1]
```

```python
ticker.fast_info["lastPrice"]
```

```python
# get latest stock price for each stock and print with stock name
for stock in stocks:
    ticker = yf.Ticker(stock)
    latest_price = ticker.history(period="1d")['Close'].iloc[-1]
    print(f"Stock: {stock}, Latest Price: {latest_price}")
```

```python
df = yf.download(
    stocks,
    start="2025-12-30",
    end="2026-01-05",
    progress=False,
)

# extract the last available close per ticker
last_closes = df["Close"].ffill().iloc[-1]

last_closes
```

```python
df = yf.download(
    stocks,
    start="2024-12-30",
    end="2025-01-05",
    group_by="ticker",
    progress=False,
    threads=True
)

out = {}

for t in stocks:
    if t in df.columns.get_level_values(0):
        out[t] = df[t]["Close"]
    else:
        out[t] = pd.Series(pd.NA, index=df.index)

close_df = pd.DataFrame(out)

close_df
```
