```python
session = requests.Session()

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) "
    "Chrome/87.0.4280.88 Safari/537.36 "
}

# Step 1: Establish NSE session (IMPORTANT)
session.get("https://www.nseindia.com", headers=headers)
```

```python
# Step 2: IPO issue page (Mainboard IPOs)
ipo_url = "https://www.nseindia.com/api/public-past-issues?&symbol=BAJAJHFL&security_type=all"
response = session.get(ipo_url, headers=headers)

data = response.json()
```
